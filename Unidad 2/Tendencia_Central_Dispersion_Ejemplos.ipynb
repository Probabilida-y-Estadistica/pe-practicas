```python
import pandas as pd
import numpy as np

# Datos a evaluar
tiempos = pd.Series(np.random.randint(25, 60, size=20))
```

```python
print("Arreglo de datos: ", tiempos)
# Calcurar
media = tiempos.mean()
mediana = tiempos.median()
moda = tiempos.mode()
rango = tiempos.max() - tiempos.min()
varianza = tiempos.var(ddof=0)
desviacion = tiempos.std(ddof=0)

print("Media: ", media)
print("Mediana: ", mediana)
print("Moda: ", moda)
print("Rango: ", rango)
print("Varianza: ", varianza)
print("Desviacion estandar: ", desviacion)
```
